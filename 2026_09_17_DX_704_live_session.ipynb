```python
import random

import matplotlib.pyplot as plt
import numpy as np
```

```python
def pick_new(n, previous_restaurants):
    return None

def plot_algorithm(n, pick_func, samples=10):
    final_totals = []
    for s in range(samples):
        sample_history = [0]
        previous_restaurants = []
        for i in range(n):
            p = pick_func(n - i, previous_restaurants)
            if p is None:
                r = random.random()
                previous_restaurants.append(r)
            else:
                r = previous_restaurants[p]

            sample_history.append(sample_history[-1] + r)

        plt.plot(sample_history)
        final_totals.append(sample_history[-1])

    plt.title(f"Algorithm = {pick_func.__name__}, Average = {np.mean(final_totals):.2f}")
    plt.xlim(0)
    plt.ylim(0)

plot_algorithm(20, pick_new)
```

```python
def try_10(n, previous_restaurants):
    if len(previous_restaurants) < 10:
        return None

    return np.argmax(previous_restaurants)

plot_algorithm(20, try_10)
```

```python
def pick_optimal(n, previous_restaurants):
    if not previous_restaurants:
        return None

    best_score = np.max(previous_restaurants)
    if best_score < np.sqrt(n) / (np.sqrt(n) + 1):
        return None

    return np.argmax(previous_restaurants)

plot_algorithm(20, pick_optimal)
```
